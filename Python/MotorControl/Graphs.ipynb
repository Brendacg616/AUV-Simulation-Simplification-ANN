```python
# Read matlab data and plot desired vs actual theta, control signal u, and PID errors 
import scipy.io
import matplotlib.pyplot as plt
import numpy as np
# Load the .mat file
mat_data = scipy.io.loadmat('./Data/motor1_diff_delta_r.mat')
# Extract motor_pout
output_data1 = mat_data['motor_pout_all']
```

```python

# Plot motor outputs from both files in diferent subplots

time = [t * 1e-3 for t in range(output_data1.shape[1])]
plt.figure(figsize=(12, 6)) 
plt.subplot(2, 1, 1)
plt.plot(time, output_data1[2000,:,0])
plt.title('Motor Control Signal Forward Swimming (Deflection Angle 0°)')
plt.xlabel('Time (s)')        
plt.ylabel('Motor Control Signal (V)')
plt.legend()
plt.grid()
plt.subplot(2, 1, 2)
plt.plot(time, output_data1[-1,:,0], color='orange')
plt.title('Motor Control Signal Turning (Deflection Angle 90°)')
plt.xlabel('Time (s)')
plt.ylabel('Motor Control Signal (V)')
plt.legend()
plt.grid()
plt.tight_layout()
plt.show()
```
